```python
import numpy as np
import pandas as pd

# Load the CSV file (make sure the filename matches what you uploaded)
df = pd.read_csv('ecommerce_sales_data (2).csv')

# Display first 5 rows
print('First 5 Rows of Dataset:')
display(df.head())

# Check total rows and columns
print('Dataset Shape (Rows, Columns):', df.shape)

# Check data types and missing values
print('\nMissing Values per Column:')
print(df.isnull().sum())

# Check and remove duplicate rows if any exist
print('\nDuplicate Rows:', df.duplicated().sum())
df = df.drop_duplicates()

# Calculate overall totals and averages using NumPy & Pandas
total_sales = np.sum(df['Sales'])
total_profit = np.sum(df['Profit'])
total_quantity = np.sum(df['Quantity'])
avg_order_sales = np.mean(df['Sales'])
profit_margin = (total_profit / total_sales) * 100

print('--- OVERALL E-COMMERCE PERFORMANCE ---')
print(f'Total Revenue (Sales): ${total_sales:,.2f}')
print(f'Total Profit:          ${total_profit:,.2f}')
print(f'Overall Profit Margin: {profit_margin:.2f}%')
print(f'Total Units Sold:      {total_quantity:,}')
print(f'Average Sale/Order:    ${avg_order_sales:,.2f}')

# 1. Performance by Category
category_summary = (
    df.groupby('Category')[['Quantity', 'Sales', 'Profit']].sum().reset_index()
)
category_summary['Profit Margin (%)'] = (
    category_summary['Profit'] / category_summary['Sales']
) * 100

print('--- SALES & PROFIT BY CATEGORY ---')
display(category_summary.sort_values(by='Sales', ascending=False))

# 2. Performance by Region
region_summary = (
    df.groupby('Region')[['Quantity', 'Sales', 'Profit']].sum().reset_index()
)
print('\n--- SALES & PROFIT BY REGION ---')
display(region_summary.sort_values(by='Profit', ascending=False))

# 1. Top 5 Best-Selling Products by Profit
top_products = (
    df.groupby('Product Name')[['Quantity', 'Sales', 'Profit']]
    .sum()
    .reset_index()
)
print('--- TOP 5 PRODUCTS BY PROFIT ---')
display(top_products.sort_values(by='Profit', ascending=False).head(5))

# 2. Monthly Sales Trend (Converting Order Date to Month-Year)
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Month_Year'] = df['Order Date'].dt.to_period('M')

monthly_trend = df.groupby('Month_Year')[['Sales', 'Profit']].sum().reset_index()
print('\n--- MONTHLY SALES & PROFIT TREND ---')
display(monthly_trend.head(12))
```
